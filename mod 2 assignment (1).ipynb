```python
import pandas as pd 
```

```python
df=pd.read_csv("Total.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
def get_employee(name):
    try:
        employee = df[df["EmployeeName"] == name]

        if employee.empty:
            return "Employee not found"

        return employee

    except Exception as e:
        return f"An error occurred: {e}"
```

```python
df["EmployeeName"].head()
```

```python
employee = get_employee("NATHANIEL FORD")
employee
```

```python
employee_dict={
    "Name": employee.iloc[0]["EmployeeName"],
    "Job Title": employee.iloc[0]["JobTitle"],
    "Base Pay": employee.iloc[0]["BasePay"],
    "Overtime Pay": employee.iloc[0]["OvertimePay"],
    "Total Pay": employee.iloc[0]["TotalPay"],
    "Year": employee.iloc[0]["Year"]
}

employee_dict    
```

```python
employee_df =pd.DataFrame([employee_dict])
employee_df.to_csv("employee_details_vanasdall.csv", index=False)
print("Employee details exported successfully!")
```
